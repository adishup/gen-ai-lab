```python
# ============================================================
# EXPERIMENT 8
# AI CODE GENERATION AND DEBUGGING
# ============================================================

import torch

from transformers import (
    AutoTokenizer,
    AutoModelForSeq2SeqLM
)


print("=" * 60)
print("EXPERIMENT 8 - AI CODE GENERATION AND DEBUGGING")
print("=" * 60)


# ============================================================
# 1. DEVICE
# ============================================================

if torch.cuda.is_available():

    device = torch.device("cuda")

    print("\nGPU Available:")
    print(torch.cuda.get_device_name(0))

else:

    device = torch.device("cpu")

    print("\nUsing CPU")


# ============================================================
# 2. LOAD CODET5
# ============================================================

model_name = "Salesforce/codet5-base"

print("\nLoading CodeT5 tokenizer...")

tokenizer = AutoTokenizer.from_pretrained(
    model_name,
    use_fast=False
)

print("Tokenizer loaded successfully.")


print("\nLoading CodeT5 model...")

model = AutoModelForSeq2SeqLM.from_pretrained(
    model_name
)

model = model.to(device)

model.eval()

print("CodeT5 model loaded successfully.")


# ============================================================
# 3. GENERATION FUNCTION
# ============================================================

def generate_text(prompt, max_new_tokens=150):

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    input_ids = inputs["input_ids"].to(device)

    attention_mask = inputs["attention_mask"].to(device)

    with torch.no_grad():

        output_ids = model.generate(
            input_ids=input_ids,
            attention_mask=attention_mask,
            max_new_tokens=max_new_tokens,
            num_beams=4,
            do_sample=False
        )

    result = tokenizer.decode(
        output_ids[0],
        skip_special_tokens=True
    )

    return result.strip()


# ============================================================
# PART A — CODE GENERATION
# ============================================================

print("\n" + "=" * 60)
print("PART A - CODE GENERATION")
print("=" * 60)


prompt = """
Generate Python code for the following task:

Write a Python function to calculate the factorial
of a number using recursion.

The function should:
- accept an integer
- return the factorial
- handle zero
- handle negative numbers

Python code:
"""


print("\nGenerating Python code...")

generated_code = generate_text(
    prompt,
    max_new_tokens=150
)


print("\nGenerated Code:")
print("-" * 60)
print(generated_code)


# ============================================================
# PART B — CODE DEBUGGING
# ============================================================

print("\n" + "=" * 60)
print("PART B - CODE DEBUGGING")
print("=" * 60)


faulty_code = """
def calculate_average(numbers):
    total = sum(numbers)
    average = total / len(number)
    return average

values = [10, 20, 30, 40]

print("Average:", calculate_average(values))
"""


debug_prompt = f"""
Find the error in this Python code and correct it.

Faulty Code:

{faulty_code}

Explain the error and provide the corrected code.
"""


print("\nAnalyzing faulty code...")

debug_result = generate_text(
    debug_prompt,
    max_new_tokens=200
)


print("\nAI Debugging Result:")
print("-" * 60)
print(debug_result)


# ============================================================
# PART C — KNOWN ERROR
# ============================================================

print("\n" + "=" * 60)
print("ERROR IDENTIFIED")
print("=" * 60)

print("""
The variable 'number' is not defined.

Incorrect:
average = total / len(number)

Correct:
average = total / len(numbers)
""")


# ============================================================
# PART D — CORRECTED CODE
# ============================================================

corrected_code = """
def calculate_average(numbers):
    total = sum(numbers)
    average = total / len(numbers)
    return average

values = [10, 20, 30, 40]

print("Average:", calculate_average(values))
"""


print("\n" + "=" * 60)
print("CORRECTED CODE")
print("=" * 60)

print(corrected_code)


# ============================================================
# PART E — EXECUTE CORRECTED CODE
# ============================================================

values = [10, 20, 30, 40]

average = sum(values) / len(values)

print("\nOutput:")
print("Values:", values)
print("Average:", average)


# ============================================================
# COMPLETION
# ============================================================

print("\n" + "=" * 60)
print("EXPERIMENT 8 COMPLETED SUCCESSFULLY")
print("=" * 60)
```
