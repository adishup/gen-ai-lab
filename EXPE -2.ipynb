```python
!pip install -q transformers torch

from transformers import pipeline

print("Loading models...")

sentiment_analyzer = pipeline(
    "sentiment-analysis",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

document_classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)

print("Models loaded successfully!")

sentiment_text = input("\nEnter text for sentiment analysis: ").strip()

if not sentiment_text:
    sentiment_text = (
        "The Generative AI workshop was extremely informative and useful."
    )

sentiment_result = sentiment_analyzer(sentiment_text)[0]

print("\nSENTIMENT ANALYSIS")
print("-" * 50)
print("Input:", sentiment_text)
print("Label:", sentiment_result["label"])
print("Confidence Score:", round(sentiment_result["score"], 3))

document = input("\nEnter a document for classification: ").strip()

if not document:
    document = (
        "Artificial Intelligence and Machine Learning are transforming "
        "industries through automation and intelligent decision-making."
    )

labels = ["Technology", "Sports", "Politics", "Entertainment"]

classification_result = document_classifier(
    document,
    candidate_labels=labels
)

print("\nDOCUMENT CLASSIFICATION")
print("-" * 50)
print("Document:", document)
print("Predicted Category:", classification_result["labels"][0])
print("Confidence Score:", round(classification_result["scores"][0], 3))
```
